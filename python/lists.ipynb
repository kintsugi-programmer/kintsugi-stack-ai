# Lists

Lists in Python represent ordered sequences of values. Here is an example of how to create them:

```python
a = [ 1, 2, 3] # A list in Python represents an ordered sequence of values. 
b = [ 'a', 'b', 'c'] # It can be of any data type. 
c = [ 1,"a", help, True] # It can be of mixed data types. 
d = [ a, b, c ]
d1 = [ [ 1, 2, 3], [ 'a', 'b', 'c'], [ 1,"a", help, True], [ a, b, c ]] # It can have a nested list. 
```

```python
primes = [2, 3, 5, 7]
```

We can put other types of things in lists:

```python
planets = ['Mercury', 'Venus', 'Earth', 'Mars', 'Jupiter', 'Saturn', 'Uranus', 'Neptune']
```

We can even make a list of lists:

```python
hands = [
    ['J', 'Q', 'K'],
    ['2', '2', '2'],
    ['6', 'A', 'K'], # (Comma after the last element is optional)
]
# (I could also have written this on one line, but it can get hard to read)
hands = [['J', 'Q', 'K'], ['2', '2', '2'], ['6', 'A', 'K']]
```

A list can contain a mix of different types of variables:

```python
my_favourite_things = [32, 'raindrops on roses', help]
# (Yes, Python's help function is *definitely* one of my favourite things)
```

```python
# Index is the way to pinpoint location out the list items.
# planets = ['Mercury', 'Venus', 'Earth', 'Mars', 'Jupiter', 'Saturn', 'Uranus', 'Neptune']
# ['Mercury', 'Venus', 'Earth', 'Mars', 'Jupiter', 'Saturn', 'Uranus', 'Neptune']
# 0         , 1      ,  2     ,  3    ,  4       ,  5       ,  6       , 7  positive indexes start from 0 from the Leftmost item from the list in Python. 
# -8        , -7     ,  -6    ,  -5   ,  -4      ,  -3      ,  -2      , -1 negative indexes start from -1 from the rightmost item from the list in Python. 
print(planets[0])
print(planets[-8])
print(planets[7])
print(planets[-1])
# Mercury
# Mercury
# Neptune
# Neptune
```

## Slicing

What are the first three planets? We can answer this question using *slicing*:

```python
# Slicing is all about taking a slice of a list.
# [ Starting inclusive, ending exclusive )
# planets = ['Mercury', 'Venus', 'Earth', 'Mars', 'Jupiter', 'Saturn', 'Uranus', 'Neptune']
print(planets[:]) # All 
print(planets[:1]) # till the first one. 
print(planets[-1:]) # Only the first one From the last 
print(planets[-2:]) #  last two elements
print(planets[-2:-1]) #  between -2 and -1
print(planets[1:3]) # between 1 to 3 
print(planets[2:-2]) # between 2 to - 2 
# ['Mercury', 'Venus', 'Earth', 'Mars', 'Jupiter', 'Saturn', 'Uranus', 'Neptune']
# ['Mercury']
# ['Neptune']
# ['Uranus', 'Neptune']
# ['Uranus']
# ['Venus', 'Earth']
# ['Earth', 'Mars', 'Jupiter', 'Saturn']
```

```python
planets[0:3]
```

`planets[0:3]` is our way of asking for the elements of `planets` starting from index 0 and continuing up to *but not including* index 3.

The starting and ending indices are both optional. If I leave out the start index, it's assumed to be 0. So I could rewrite the expression above as:

```python
planets[:3]
```

If I leave out the end index, it's assumed to be the length of the list.

```python
planets[3:]
```

i.e. the expression above means "give me all the planets from index 3 onward".

We can also use negative indices when slicing:

```python
# All the planets except the first and last
planets[1:-1]
```

```python
# The last 3 planets
planets[-3:]
```

## Changing lists

Lists are "mutable", meaning they can be modified "in place".

One way to modify a list is to assign to an index or slice expression.

For example, let's say we want to rename Mars:

```python
# Lists are mutable(Inside stuff can be changed) In Python 
planets[-1] = "AfghaniMomo"
print(planets)
# ['Mercury', 'Venus', 'Earth', 'Mars', 'Jupiter', 'Saturn', 'Uranus', 'AfghaniMomo']
planets[1:3] = [ 1, 2 ]
print(planets)
# ['Mercury', 1, 2, 'Mars', 'Jupiter', 'Saturn', 'Uranus', 'AfghaniMomo']
```

```python
planets[3] = 'Malacandra'
planets
```

Hm, that's quite a mouthful. Let's compensate by shortening the names of the first 3 planets.

```python
planets[:3] = ['Mur', 'Vee', 'Ur']
print(planets)
# That was silly. Let's give them back their old names
planets[:4] = ['Mercury', 'Venus', 'Earth', 'Mars',]
```

## List functions

Python has several useful functions for working with lists.

`len` gives the length of a list:

```python
# Python has inbuilt useful functions for working with lists. 
planets = ['Mercury', 'Venus', 'Earth', 'Mars', 'Jupiter', 'Saturn', 'Uranus', 'Neptune']
print(len(planets))
print(sorted(planets)) # it returns sorted version, not changes the list
primes = [2, 3, 5, 7]
print(sum(primes))
# 8
# ['Earth', 'Jupiter', 'Mars', 'Mercury', 'Neptune', 'Saturn', 'Uranus', 'Venus']
# 17
print(max(primes))
print(min(primes))
# 7
# 2
```

```python
# How many planets are there?
len(planets)
```

`sorted` returns a sorted version of a list:

```python
# The planets sorted in alphabetical order
sorted(planets)
```

`sum` does what you might expect:

```python
primes = [2, 3, 5, 7]
sum(primes)
```

We've previously used the `min` and `max` to get the minimum or maximum of several arguments. But we can also pass in a single list argument.

```python
max(primes)
```

## Interlude: objects

I've used the term 'object' a lot so far - you may have even read that *everything* in Python is an object. What does that mean?

In short, objects carry some things around with them. You access that stuff using Python's dot syntax.

For example, numbers in Python carry around an associated variable called `imag` representing their imaginary part. (You'll probably never need to use this unless you're doing some very weird math.)

```python
# Objects are those things which carry some things around with them. 
# In Python, everything is an object. 
# A function that is attached to an object is called a method. 
# Non-functional things attached to the object are called attributes. 
x = 13
print(x.real)
print(x.imag) #As it's a real number, its imaginary number would be zero. 
# 13
# 0
c = 1 + 2j
print(c.real) # tells the real part of the number 
print(c.imag) #  tells the imaginary part of the number 
# 1.0
# 2.0
print(x.bit_length) # <built-in method bit_length of int object at 0xa427e8>
print(x.bit_length())#  To call a functional method, we need to use round brackets
# 4
print(help(x.bit_length)) # doing a help function parameter, we don't need to write round brackets  but as this function-specific one is not an all-rounder universal built-in function( like max or len ), we have to use a list for asking about this function's metadata 
# Help on built-in function bit_length:
# bit_length() method of builtins.int instance
#     Number of bits necessary to represent self in binary.
#     >>> bin(37)
#     '0b100101'
#     >>> (37).bit_length()
#     6
# None
```

```python
x = 12
# x is a real number, so its imaginary part is 0.
print(x.imag)
# Here's how to make a complex number, in case you've ever been curious:
c = 12 + 3j
print(c.imag)
```

The things an object carries around can also include functions. A function attached to an object is called a **method**. (Non-function things attached to an object, such as `imag`, are called *attributes*).

For example, numbers have a method called `bit_length`. Again, we access it using dot syntax:

```python
x.bit_length
```

To actually call it, we add parentheses:

```python
x.bit_length()
```

> **Aside:** You've actually been calling methods already if you've been doing the exercises. In the exercise notebooks `q1`, `q2`, `q3`, etc. are all objects which have methods called `check`, `hint`, and `solution`.

In the same way that we can pass functions to the `help` function (e.g. `help(max)`), we can also pass in methods:

```python
help(x.bit_length)
```

## Tuples

Tuples are almost exactly the same as lists. They differ in just two ways.

**1:** The syntax for creating them uses parentheses instead of square brackets

```python
# Tuple uses parentheses instead of square brackets. 
# They cannot be modified once created. They are immutable. 
t = (1,2,3,4)
t1 = 1,2,3,4 # same way
print(t)
print(t1)
# (1, 2, 3, 4)
# (1, 2, 3, 4)

# t[1]=100 # This won't work and will throw an error. 
# ---------------------------------------------------------------------------
# TypeError                                 Traceback (most recent call last)
# /tmp/ipykernel_40/3947249233.py in <cell line: 0>()
#       5 print(t)
#       6 print(t1)
# ----> 7 t[1]=100
# TypeError: 'tuple' object does not support item assignment

x = 11.22
Numerator, denominator = x.as_integer_ratio() # this method will return a tuple splitting numerator and denominator  
print(Numerator)
print(denominator)
# 6316298477387121
# 562949953421312

# Computers store floating-point numbers in binary (base 2), not decimal (base 10).
# and 6316298477387121 / 562949953421312 = 11.22
print(Numerator/denominator)
# 11.22

# Classic Python stupid trick to swap two variables. 
a = 10
b = 20
print(a,b)
a,b=b,a
print(a,b)
# 10 20
# 20 10
```

```python
t = (1, 2, 3)
```

```python
t = 1, 2, 3 # equivalent to above
t
```

**2:** They cannot be modified (they are *immutable*).

```python
t[0] = 100
```

Tuples are often used for functions that have multiple return values.

For example, the ``as_integer_ratio()`` method of float objects returns a numerator and a denominator in the form of a tuple:

```python
x = 0.125
x.as_integer_ratio()
```

These multiple return values can be individually assigned as follows:

```python
numerator, denominator = x.as_integer_ratio()
print(numerator / denominator)
```

Finally we have some insight into the classic Stupid Python Trick™ for swapping two variables!

```python
a = 1
b = 0
a, b = b, a
print(a, b)
```
